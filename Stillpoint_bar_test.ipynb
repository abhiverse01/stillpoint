```python


import random
import time

from stillpoint import stillpoint, DAWN, DUSK, MOSS


def fake_step(base_delay=0.03):
    """Pretend to do some work — a training step, a download chunk, whatever."""
    time.sleep(base_delay + random.uniform(-0.01, 0.01))


def main():
    print("\nstillpoint demo — a calm progress bar for real work\n")

    # --- Epoch 1: default DAWN gradient, full bar mode ---
    with stillpoint.track(total=60, label="epoch 1/3", gradient=DAWN) as bar:
        for _ in range(60):
            fake_step()
            bar.advance()

    # --- A pause between epochs: indeterminate wait, breathing glyph ---
    with stillpoint.breathe(label="saving checkpoint"):
        time.sleep(1.5)

    # --- Epoch 2: DUSK gradient ---
    with stillpoint.track(total=60, label="epoch 2/3", gradient=DUSK) as bar:
        for _ in range(60):
            fake_step()
            bar.advance()

    with stillpoint.breathe(label="saving checkpoint"):
        time.sleep(1.5)

    # --- Epoch 3: MOSS gradient, using the iterator form instead ---
    for _ in stillpoint(range(60), label="epoch 3/3", gradient=MOSS):
        fake_step()

    # --- Quiet mode: a single evolving glyph, for validation pass ---
    for _ in stillpoint(range(40), label="validating", mode="quiet"):
        fake_step(0.02)

    print("\ntraining complete.\n")


if __name__ == "__main__":
    main()
```
